```python
import pandas as pd

# Charger les données du fichier CSV
def load_cocktails():
    df = pd.read_csv("all_cocktails.csv")
    # Remplacer les valeurs manquantes par des chaînes vides
    df.fillna("", inplace=True)
    return df

# Appeler la fonction pour charger les données au démarrage
df = load_cocktails()
```

```python
# Fonction pour recommander des cocktails en fonction des ingrédients
def recommend_cocktails(ingredients):
    recommended = []
    for _, row in df.iterrows():
        cocktail_ingredients = [row[f"strIngredient{i}"] for i in range(1, 16) if row[f"strIngredient{i}"]]
        if all(ingredient.lower() in [i.lower() for i in cocktail_ingredients] for ingredient in ingredients):
            recommended.append(row["strDrink"])
    return recommended
```

```python
# Fonction pour obtenir la recette d'un cocktail
def get_recipe(cocktail_name):
    row = df[df['strDrink'].str.lower() == cocktail_name.lower()]
    if not row.empty:
        instructions = row["strInstructions"].values[0]
        ingredients = [(row[f"strMeasure{i}"].values[0], row[f"strIngredient{i}"].values[0]) 
                       for i in range(1, 16) 
                       if row[f"strIngredient{i}"].values[0]]
        return {
            "name": row["strDrink"].values[0],
            "instructions": instructions,
            "ingredients": [f"{measure} {ingredient}" for measure, ingredient in ingredients if ingredient]
        }
    return None
```

```python
def chatbot():
    print("Bienvenue dans le chatbot de recommandation de cocktails !")
    print("Je peux vous aider à trouver des cocktails ou à obtenir des recettes.")
    
    while True:
        user_input = input("\nQue puis-je faire pour vous ? (ex: 'recommander', 'recette', 'quitter') : ").strip().lower()
        
        if user_input == "quitter":
            print("Merci d'avoir utilisé le chatbot. À bientôt !")
            break
        
        elif user_input == "recommander":
            ingredients_input = input("Quels ingrédients avez-vous ? (séparés par des virgules) : ")
            ingredients = [ingredient.strip().lower() for ingredient in ingredients_input.split(",")]
            recommendations = recommend_cocktails(ingredients)
            
            if recommendations:
                print(f"Voici quelques cocktails que vous pouvez préparer : {', '.join(recommendations)}")
            else:
                print("Désolé, je ne trouve aucun cocktail correspondant à vos ingrédients.")
        
        elif user_input == "recette":
            cocktail_name = input("De quel cocktail voulez-vous la recette ? : ").strip().lower()
            recipe = get_recipe(cocktail_name)
            
            if recipe:
                print(f"\n{recipe['name']} :")
                print(f"Ingrédients : {', '.join(recipe['ingredients'])}")
                print(f"Instructions : {recipe['instructions']}")
            else:
                print("Désolé, je ne connais pas ce cocktail.")
        
        else:
            print("Je ne comprends pas votre demande. Essayez 'recommander', 'recette' ou 'quitter'.")
```

```python
if __name__ == "__main__":
    chatbot()
```
